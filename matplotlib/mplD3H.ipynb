```python
import matplotlib.pyplot as plt
months = [1, 2, 3, 4]
product_a = [100, 150, 200, 250]
product_b = [120, 170, 210, 260]
fig,ax=plt.subplots()
ax.plot(months, product_a, color='green', marker='o', linestyle=':', label='Product A')
ax.plot(months, product_b, color='red', marker='s', linestyle='--', label='Product B')
plt.legend()
plt.grid(True)
plt.ylim(0,300)
plt.xticks([1, 2, 3, 4], [ "Jan", "Feb", "Mar", "Apr"])
plt.title("Monthly Product Sales Comparison", fontsize=14, color="blue")
plt.xlabel("Months", fontsize=12)
plt.ylabel("Sales(unit)", fontsize=12)
plt.show()
```
