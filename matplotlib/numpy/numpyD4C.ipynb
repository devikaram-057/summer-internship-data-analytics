```python
import numpy as np
marks=np.array([35, 55, 75, 40, 90])
print("->Marks:",marks)
print("\n->np.where:",np.where(marks>=40,"Pass","Fail"))
conditions = [marks<40, marks<60,marks<80]
choices = ['Fail', 'C','B']
result = np.select(conditions, choices, default='A')
print("\n->np.select:",result)
d_marks=np.array([50, 60, 50, 70, 60, 80])
print("\n->np.unique:",np.unique(d_marks))
print("\n->Any student has failed(np.any):",np.any(marks<40))
print("\n->All students passed(np.all):",np.all(marks>=40))
```
