```python
import numpy as np
import pandas as pd
marks = np.array([45, 55, 60, 70, 85])
marks_with_grace = marks + 5
print("Marks after adding grace:", marks_with_grace)
min_val = marks_with_grace.min()
max_val = marks_with_grace.max()
normalized_marks = (marks_with_grace - min_val) / (max_val - min_val)
print("\nNormalized Marks:", normalized_marks)
np.savetxt("normalized_marks.csv", normalized_marks, delimiter=",")
print("\nNormalized marks saved to 'normalized_marks.csv'")
loaded_marks = np.genfromtxt("normalized_marks.csv", delimiter=",")
print("\nLoaded Marks:", loaded_marks)
df = pd.DataFrame(loaded_marks, columns=["Normalized Marks"])
print("\nFinal DataFrame:")
print(df)
```
