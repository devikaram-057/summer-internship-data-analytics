```python
import numpy as np
salaries=np.array([35000, 42000, np.nan, 50000, np.inf, 38000, np.nan])
print("=>Salary Details:\n")
print("->Original data:",salaries)
salaries[np.isnan(salaries)] =30000
salaries[np.isinf(salaries)] =30000
print("->Clear data:",salaries)
arr=np.array([10, 20, 30, 40])
copy_arr = arr.copy()
view_arr = arr.view()
view_arr[0] =999
print("\n=>copy() vs view():\n")
print("->Original Array:",arr)
print("->Copy Array:",copy_arr)
print("->View Array:",view_arr)
#Both arrays point to the same memory location, 
#so writing new values using view() also changes the original array automatically.
#The copy() method creates a full separate array which increases memory usage.  
```
