```python
import numpy as np
sens_readings=np.array([25.5, 26.0, np.nan, 27.8, np.inf, 26.4])
print("Sensor Data Collecting Details:\n")
print("->Original Data:",sens_readings)
print("->Boolean masks for NaN and inf:")
print(np.isnan(sens_readings))
print(np.isinf(sens_readings))
print("->Cleaned data after replacement:")
sens_readings[np.isnan(sens_readings)] = 0
sens_readings[np.isinf(sens_readings)] = 0
print(sens_readings)
```
