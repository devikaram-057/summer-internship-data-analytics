```python
import matplotlib.pyplot as plt
import numpy as np
import datetime
dates = [datetime.date(2025, 1, d) for d in range(1, 6)]
sales = [100, 120, 150, 180, 220]
plt.plot(dates,sales)
plt.title(" Date-Based Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.show()
dates_2= [datetime.date(2025, 1, d) for d in range(1, 5)]
sales_2= [100, 200, 300, 400]
profit=[20, 50, 90, 150]
fig, ax1 = plt.subplots()
ax2 = ax1.twinx()
ax1.plot(dates_2,sales_2)
ax2.plot(dates_2,profit)
plt.xlabel("Date")
ax1.set_ylabel("Sales")
ax2.set_ylabel("Profit")
plt.show()
x = [1, 2, 3, 4]
y = [10, 100, 1000, 10000]
plt.plot(x, y)
plt.yscale("log")
plt.title("Exponential Growth Analysis")
plt.show()
Product_A=[30, 40, 50, 60]
Product_B=[20, 30, 40, 50]
plt.stackplot(Product_A,Product_B)
plt.title("Stacked Sales Contribution")
plt.xlabel("Product_A")
plt.ylabel("Product_B")
plt.show()
```
