```python
import matplotlib.pyplot as plt
import numpy as np
days = [1, 2, 3, 4, 5]
sales = [100, 150, np.nan, 200, 250]
plt.plot(days,sales)
plt.scatter([5], [250], color="red")
plt.annotate("Highest value",(5,250),textcoords="offset points",xytext=(0,10), arrowprops=dict(arrowstyle="->"))
plt.title("Daily Sales Report")
plt.xlabel("Day")
plt.ylabel("Sales Amount")
plt.show()
```
