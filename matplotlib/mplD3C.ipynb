```python
import matplotlib.pyplot as plt
months = [1, 2, 3, 4]
product_a = [200, 250, 300, 350]
product_b = [180, 220, 260, 310]
fig,ax=plt.subplots()
ax.plot(months, product_a, color='green', marker='o', linestyle=':', label='Product A')
ax.plot(months, product_b, color='red', marker='s', linestyle='--', label='Product B')
ax.set_xlabel("Month")
ax.set_ylabel("Sales")
plt.show()
ax.set_xlabel("Month")
ax.set_ylabel("Sales")
ax.legend()
plt.show()
fig2, (ax1, ax2) = plt.subplots(2, 1, sharex=True)
ax1.plot(months, product_a, color='green', marker='o')
ax1.set_title("Product A Sales")
ax1.set_ylabel("Sales")
ax2.plot(months, product_b, color='red', marker='s')
ax2.set_title("Product B Sales")
ax2.set_xlabel("Month")
ax2.set_ylabel("Sales")
plt.show()
```
