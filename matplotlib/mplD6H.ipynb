```python
import matplotlib.pyplot as plt
import numpy as np
months = [1, 2, 3, 4]
sales_2024 = [200, 260, 300, 380] 
sales_2025 = [220, 280, 330, 420]
plt.style.use('seaborn-v0_8')
plt.figure()
plt.plot(months, sales_2024, label="Sales 2024", alpha=0.7)
plt.plot(months, sales_2025, label="Sales 2025", alpha=0.7)
plt.title(" Sales Comparison (2024 vs 2025)")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)
plt.savefig(" sales_report.png", dpi=200)
plt.show()
np.random.seed(0)
x = np.random.rand(50)
y = np.random.rand(50)
intensity = np.random.rand(50)
plt.figure()
scatter = plt.scatter(x, y, c=intensity, cmap='viridis')
plt.colorbar(scatter)
plt.title("Sales Intensity Distribution")
plt.show()
plt.figure(figsize=(8, 4))
plt.plot([1, 2, 3], [10, 20, 30])
plt.tight_layout()
plt.savefig("report_chart.png", dpi=300)
plt.savefig("report_chart.svg", dpi=300)
plt.show()
```
