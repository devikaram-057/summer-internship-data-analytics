```python
import numpy as np
scores=np.array([25, 45, 65, 85, 95])
print("->Scores:",scores)
print("\n->np.where:",np.where(scores<40,"Poor","Good"))
conditions = [scores<40, scores<70,scores<90]
choices = ['Poor', 'Average','Good']
result = np.select(conditions, choices, default='Excellent')
print("\n->np.select:",result)
d_marks=np.array([101, 102, 101, 103, 102, 104])
print("\n->np.unique:",np.unique(d_marks))
print("\n->Any high scorer(np.any):",np.any(scores<90))
print("\n->All students above minimum (np.all):",np.all(scores>=30))
```
