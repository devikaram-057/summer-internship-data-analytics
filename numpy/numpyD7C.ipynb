```python
import numpy as np
import sys
temps_c = np.array([28, 30, 32, 35, 33])
temps_f_loop = []
for c in temps_c:
    f = (c * 9/5) + 32
    temps_f_loop.append(f)
print("Method 1 (For Loop) - Fahrenheit:")
print(temps_f_loop)
temps_f_vectorized = (temps_c * 9/5) + 32
print("\nMethod 2 (Vectorized) - Fahrenheit:")
print(temps_f_vectorized)
temps_list = [28, 30, 32, 35, 33]
print("\nMemory Usage:")
print("Python List:", sys.getsizeof(temps_list), "bytes")
print("NumPy Array:", temps_c.nbytes, "bytes")
np.save("temps_fahrenheit.npy", temps_f_vectorized)
print("\nVectorized result saved to 'temps_fahrenheit.npy'")
loaded_data = np.load("temps_fahrenheit.npy")
print("\nLoaded Data from file:")
print(loaded_data)
```
