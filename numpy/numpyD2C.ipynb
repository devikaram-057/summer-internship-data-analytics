```python
import numpy as np
arr_rand=np.random.rand(5)
arr_randn=np.random.randn(5)
print("->Random values representing normalized scores:",arr_rand)
print("\n->Random values from a normal distribution:",arr_randn)
arr_int=np.random.randint(35, 100, (3,4))
print("\n->3×4 array of random integers between 35 and 100:\n",arr_int)
student_IDs=np.array([101, 102, 103, 104, 105])
print("\n->Randomly select 2 students:",np.random.choice(student_IDs, size=2))
np.random.seed(10)
print("\n->Generate 3 random values using seed:",np.random.choice(student_IDs, size=3))
arr=np.array([10, 20, 30, 40, 50, 60])
print("\n->3rd element:",arr[2])
print("\n->Elements from index 1 to 4:",arr[1:5])
print("\n->Last element:",arr[-1])
print("\n->Every second element:",arr[1::2])
print("\n->Reversed array:",arr[::-1])
```
