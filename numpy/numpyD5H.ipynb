```python
import numpy as np
daily_sales=np.array([250.75, 300.40, 275.60, 400.90, 150.30, 500.50, 350.80])
print("Daily sales Analysis\n")
print("->Total sales:",(np.sum(daily_sales)))
print("->Average sales:",(np.mean(daily_sales)))
print("->Median sales:",(np.median(daily_sales)))
print("->Sales variation:",(np.std(daily_sales)))
print("->Minimum sale:",(np.min(daily_sales)))
print("->Maximum sale:",(np.max(daily_sales)))
print("->floor() to sales:",np.floor(daily_sales))
print("->ceil() to sales:",np.ceil(daily_sales))
print("->round() to sales:",np.round(daily_sales))
print("->Exponential growth:",np.exp(daily_sales))
print("->Cumulative sales:",np.cumsum(daily_sales))
A=np.array([[2, 1],[3, 4]])
B= np.array([[1, 2],[5, 6]])
print("\nMatrix Operations:\n")
print("A=\n",A)
print("B=\n",B)
print("->Matrix multiplication:\n",A@B)
print("->Determinant of matrix A:",np.linalg.det(A))
print("->Inverse of matrix A:\n",np.linalg.inv(A))
values, vectors = np.linalg.eig(A)
print("Eigenvalues:", values)
print("Eigenvectors:\n", vectors)
a=np.array([[2,1],[3,4]])
b=np.array([5,11])
x = np.linalg.solve(a, b)
print("->Linear equation solution:",x)
```
