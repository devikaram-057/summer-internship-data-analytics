```python
import numpy as np
st_marks=np.array([78, 85, 62, 90, 55, 88])
print("Student Marks Analysis\n")
print("->Total marks:",(np.sum(st_marks)))
print("\n->Average marks:",(np.mean(st_marks)))
print("\n->Median:",(np.median(st_marks)))
print("\n->Standard deviation:",(np.std(st_marks)))
print("\n->Minimum mark:",(np.min(st_marks)))
print("\n->Maximum mark:",(np.max(st_marks)))
grace_mark=st_marks+5
bonus_mark=grace_mark*1.1
print("\n->Added 5 Grace marks:",grace_mark)
print("\n->Added 10% bonus marks:",bonus_mark)
```
