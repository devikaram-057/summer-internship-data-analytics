```python
import numpy as np
st_att=np.random.rand(10)
print("\n->10 random values between 0 and 1:\n",st_att)
present=st_att>0.5
absent=st_att<=0.5
print("\n->Present values using Boolean indexing:",st_att[present])
scores=np.random.randint(0,100,15)
print("Scores:",scores)
print("\n->Scores greater than 60:",scores[scores>60])
print("\n->Scores  divisible by 5:",scores[scores%5==0])
arr=np.array( [100, 200, 300, 400, 500])
indices=[0,2,4]
print("\n->1st, 3rd and 5th elements using fancy indexing:",arr[indices])
arr_int=np.random.randint(1,50,(4,4))
print("\n->4x4 matrix:\n",arr_int)
print("\n->Entire 2nd row:",arr_int[1,:])
print("\n->Entire 3rd column:",arr_int[:,2])
uniform_vals = np.random.uniform(0, 1, size=5)
normal_vals = np.random.normal(0, 1, size=5)
print("->\n5 values using np.random.uniform(0, 1):",uniform_vals)
print("->\n5 values using np.random.normal(0, 1):",normal_vals)
```
