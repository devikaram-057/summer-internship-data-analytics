```python
import seaborn as sns
import matplotlib.pyplot as plt 
tips = sns.load_dataset("tips")
print("First 5 rows of the dataset:\n",tips.head())
sns.scatterplot(data=tips,x="total_bill", y="tip")
plt.title("Scatter plot:total_bill vs tip")
plt.show()
sns.boxplot(data=tips, x="day", y="total_bill")
plt.title("Box plot: Distribution of total_bill by day")
plt.show()
sns.histplot(data=tips, x="total_bill")
plt.title("Histogram: Distribution of total_bill")
plt.show()
sns.kdeplot(data=tips, x="total_bill")
plt.title("KDE plot: Smoothed distribution of total_bill")
plt.show()
```
