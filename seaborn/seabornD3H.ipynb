```python
import seaborn as sns
import matplotlib.pyplot as plt 
tips = sns.load_dataset("tips")
sns.set_style("whitegrid")
sns.set_context("talk")
sns.countplot(data=tips,x="size",hue="day",palette="Set2")
plt.title("Count plot of customers by day")
plt.show()
flights = sns.load_dataset("flights")
sns.lineplot(data=flights, x="year", y="passengers",palette="blues")
plt.title("Line plot of passengers by year")
plt.show()
custom_colors = ["#1f77b4", "#ff7f0e", "#2ca02c"]
sns.countplot(data=tips, x="smoker", palette=custom_colors)
plt.title("Count plot of smoker")
plt.show()
sns.scatterplot(data=tips,x="total_bill",y="tip",hue="time",palette="coolwarm")
plt.title("Scatter plot of total_bill vs tip")
plt.show()
```
