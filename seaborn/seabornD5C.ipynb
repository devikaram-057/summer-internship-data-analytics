```python
#Scenario: Employee Data Visual Analysis

import seaborn as sns
import matplotlib.pyplot as plt 
import pandas as pd
df=pd.read_csv("sample_set.csv")
sns.pairplot(df)
plt.title("Pairplot for all numerical columns")
plt.show()
sns.heatmap(df, cmap="coolwarm")
plt.title("Correlation heatmap")
plt.savefig("employee_correlation.png", dpi=300)
plt.show()
```
