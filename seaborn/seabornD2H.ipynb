```python
import seaborn as sns
import matplotlib.pyplot as plt 
tips = sns.load_dataset("tips")
sns.relplot(data=tips, x="total_bill", y="tip", col="time")
plt.title("Figure-level plot (relplot) of total_bill vs tip split by time")
plt.show()
sns.violinplot(data=tips,x="day",y="total_bill")
plt.title("Violin plot of tip_percentage by gender")
plt.show()
sns.stripplot(data=tips, x="day", y="total_bill")
plt.title("Strip plot showing individual bills per day")
plt.show()
sns.swarmplot(data=tips, x="day", y="total_bill")
plt.title("Swarm plot for better point separation")
plt.show()
sns.displot(data=tips, x="total_bill", col="time", kde=True)
plt.title("Displot of total_bill split by time with KDE")
plt.show()
sns.ecdfplot(data=tips, x="total_bill")
plt.title("ECDF plot for total_bill")
plt.show()
sns.regplot(data=tips, x="total_bill", y="tip")
plt.title("Regplot showing trend between total_bill and tip")
plt.show()
sns.lmplot(data=tips, x="total_bill", y="tip", hue="sex")
plt.title("lmplot split by sex")
plt.show()
```
