```python
import seaborn as sns
import matplotlib.pyplot as plt 
import pandas as pd
sns.set_palette("pastel")
tips=sns.load_dataset("tips")
tips['tip_percentage']=(tips['tip']/tips['total_bill'])*100
print("First 5 rows of the dataset:\n",tips.head())
sns.scatterplot(data="tips",x="total_bill", y="tip")
plt.title("Scatter plot: total_bill vs tip")
plt.show()
sns.boxplot(data=tips, x="sex", y="total_bill")
plt.title("Boxplot of total_bill by gender")
plt.show()
sns.violinplot(data=tips,x="sex",y="tip_percentage")
plt.title("Violin plot of tip_percentage by gender")
plt.show()
sns.relplot(data=tips, x="total_bill", y="tip", col="time")
plt.title("Figure-level plot (relplot) of total_bill vs tip split by time")
plt.show()
```
