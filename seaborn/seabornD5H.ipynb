```python
#Scenario: Dashboard-Ready Employee Insights

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv("sample_set.csv")
sns.scatterplot(df,x="Age",y="Salary",hue="Hours_Worked")
plt.title("Age vs Salary by Hours worked")
plt.title("Correlation Heatmap")
plt.savefig("salary_dashboard.svg",dpi=300)
plt.show()
sns.heatmap(df)
plt.show()
data = np.loadtxt("sample_set.csv", delimiter=",", skiprows=1, usecols=1)
sns.histplot(data)
plt.show()
```
